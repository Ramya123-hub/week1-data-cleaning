```python
import pandas as pd
import random

random.seed(42)

# Sample applicant information
names = [
    "Ramya Pulugurtha", "Akhil Kumar", "Sneha Reddy",
    "Rahul Sharma", "Priya Singh", "Divya Rao",
    "Kiran Kumar", "Anjali Das", "Vikram Reddy",
    "Meena Patel", "Arjun Rao", "Kavya Sharma",
    "Rohit Kumar", "Pooja Reddy", "Sai Krishna",
    "Nisha Patel", "Manoj Das", "Sravani Rao",
    "Harsha Vardhan", "Deepika Singh"
]

domains = [
    "Web Development", "web dev", "WEB DEV",
    "Data Science", "DATA SCIENCE",
    "AI", "ai", "Artificial Intelligence",
    "Cyber Security", "CYBER SECURITY",
    "Cloud Computing"
]

universities = [
    "BABA Institute", "ABC University",
    "XYZ College", "KL University",
    "Andhra University", None
]

statuses = [
    "Under Review", "selected", "SELECTED",
    "Rejected", "Under Review", None
]

rows = []

for i in range(100):
    name = random.choice(names)
    email = f"applicant{i+1}@example.com"
    phone = f"98765{10000+i}"
    domain = random.choice(domains)
    university = random.choice(universities)
    date = random.choice([
        "2026-05-12", "13/05/2026",
        "2026-05-14", "15/05/2026"
    ])
    status = random.choice(statuses)

    # Introduce realistic data-quality problems
    if i % 13 == 0:
        name = " " + name + " "

    if i % 11 == 0:
        email = None

    if i % 10 == 0:
        phone = phone[:5] + "-" + phone[5:]

    if i % 12 == 0:
        university = None

    if i % 15 == 0:
        status = None

    rows.append([
        name, email, phone, domain,
        university, date, status
    ])

df = pd.DataFrame(rows, columns=[
    "Applicant Name",
    "Email",
    "Phone",
    "Domain Applied",
    "University",
    "Application Date",
    "Status"
])

# Introduce two exact duplicate rows
df = pd.concat([df, df.iloc[[1, 2]]], ignore_index=True)

# Save the original practice dataset
df.to_csv("applicants.csv", index=False)

print("Dataset created successfully!")
print("Rows and columns:", df.shape)

display(df.head(10))
```

Initial Data Exploration
The dataset contains 102 rows and 7 columns. I used Pandas functions to inspect the first five records, examine column data types, count missing values, identify exact duplicate rows, and inspect unique domain values. The initial exploration revealed missing information, duplicate records, inconsistent text formatting, and application dates stored as text. These issues will be addressed during the data-cleaning process.


```python
# Check missing values before cleaning
print("Missing values before cleaning:")
print(df.isnull().sum())

# Remove rows where applicant name is missing
df = df.dropna(subset=["Applicant Name"])

# Fill missing email addresses with a placeholder
df["Email"] = df["Email"].fillna("Not Provided")

# Fill missing university names
df["University"] = df["University"].fillna("Not Specified")

# Fill missing application statuses
df["Status"] = df["Status"].fillna("Under Review")

# Check missing values after cleaning
print("\nMissing values after cleaning:")
print(df.isnull().sum())
```

Missing Value Handling
Missing applicant names were handled by removing those rows because names are important for identifying applicants. Missing email addresses were replaced with Not Provided, and missing university names were replaced with Not Specified. Missing application statuses were filled with Under Review, assuming those applications had not yet been processed


```python
# Check exact duplicate rows
print("Exact duplicates before removal:", df.duplicated().sum())

# Remove exact duplicate rows
before = len(df)
df = df.drop_duplicates()
print("Exact duplicate rows removed:", before - len(df))

# Check duplicate email addresses
valid_emails = df["Email"] != "Not Provided"

print(
    "Duplicate valid email addresses:",
    df.loc[valid_emails, "Email"].duplicated().sum()
)

# Remove repeated applications with the same valid email
before = len(df)

with_email = df[valid_emails].drop_duplicates(
    subset=["Email"],
    keep="first"
)

without_email = df[~valid_emails]

df = pd.concat(
    [with_email, without_email],
    ignore_index=True
)

print("Duplicate-email rows removed:", before - len(df))
print("Rows remaining:", len(df))
```

Duplicate Removal
I checked for exact duplicate rows using duplicated() and removed them using drop_duplicates(). I also checked for repeated applications using valid email addresses. When duplicate email addresses were found, I retained the first record. Applicants without an email address were handled separately to avoid incorrectly treating all of them as the same person.


```python
# Remove extra spaces from text columns
text_columns = [
    "Applicant Name",
    "Email",
    "Phone",
    "Domain Applied",
    "University",
    "Status"
]

for column in text_columns:
    df[column] = df[column].astype("string").str.strip()

# Standardize names and emails
df["Applicant Name"] = df["Applicant Name"].str.title()
df["Email"] = df["Email"].str.lower()

# Standardize domain labels
df["Domain Applied"] = df["Domain Applied"].str.lower()

domain_mapping = {
    "web dev": "Web Development",
    "web development": "Web Development",
    "webdev": "Web Development",
    "data science": "Data Science",
    "ai": "AI",
    "artificial intelligence": "AI",
    "cyber security": "Cyber Security",
    "cloud computing": "Cloud Computing"
}

df["Domain Applied"] = df["Domain Applied"].map(domain_mapping)

# Standardize application statuses
df["Status"] = df["Status"].str.title()

# Check results
print("Unique domains:")
print(df["Domain Applied"].unique())

print("\nUnique statuses:")
print(df["Status"].unique())
```

Text Standardization
I removed unnecessary whitespace, standardized applicant names and email addresses, mapped inconsistent domain labels to standard categories, and converted application statuses to a consistent format. I inspected the unique domain values to identify any unmapped entries.


```python
# Convert application dates into datetime format
df["Application Date"] = pd.to_datetime(
    df["Application Date"],
    errors="coerce",
    dayfirst=True
)

# Remove dashes and spaces from phone numbers
df["Phone"] = (
    df["Phone"]
    .str.replace("-", "", regex=False)
    .str.replace(" ", "", regex=False)
)

# Verify the data types
df.info()

print("\nInvalid or missing dates:")
print(df["Application Date"].isnull().sum())
```

Data Type Correction
I converted application dates from text to datetime values using pd.to_datetime(). Invalid dates were converted to NaT for investigation. Phone numbers were kept as text, with spaces and dashes removed for consistency.

```python
# Step 1: Handle missing values

print("Missing values before cleaning:")
print(df.isnull().sum())

# Remove rows where applicant name is missing
df = df.dropna(subset=["Applicant Name"])

# Fill missing values
df["Email"] = df["Email"].fillna("Not Provided")
df["University"] = df["University"].fillna("Not Specified")
df["Status"] = df["Status"].fillna("Under Review")

print("\nMissing values after cleaning:")
print(df.isnull().sum())

print("\nDataset size:", df.shape)
```

```python
# Step 2: Remove duplicate records

print("Exact duplicates before removal:", df.duplicated().sum())

before = len(df)

# Remove completely identical rows
df = df.drop_duplicates()

print("Duplicate rows removed:", before - len(df))

# Check for duplicate valid email addresses
valid_emails = df["Email"] != "Not Provided"

print(
    "Duplicate valid emails:",
    df.loc[valid_emails, "Email"].duplicated().sum()
)

# Keep the first record for each valid email
with_email = df[valid_emails].drop_duplicates(
    subset=["Email"],
    keep="first"
)

without_email = df[~valid_emails]

df = pd.concat(
    [with_email, without_email],
    ignore_index=True
)

print("Rows remaining:", len(df))
```

```python
# Step 3: Standardize text

# Remove extra spaces from text columns
text_columns = [
    "Applicant Name", "Email", "Phone",
    "Domain Applied", "University", "Status"
]

for column in text_columns:
    df[column] = df[column].astype("string").str.strip()

# Standardize names and emails
df["Applicant Name"] = df["Applicant Name"].str.title()
df["Email"] = df["Email"].str.lower()

# Standardize application domains
df["Domain Applied"] = df["Domain Applied"].str.lower()

domain_mapping = {
    "web dev": "Web Development",
    "web development": "Web Development",
    "webdev": "Web Development",
    "data science": "Data Science",
    "ai": "AI",
    "artificial intelligence": "AI",
    "cyber security": "Cyber Security",
    "cloud computing": "Cloud Computing"
}

df["Domain Applied"] = df["Domain Applied"].map(domain_mapping)

# Standardize application statuses
df["Status"] = df["Status"].str.title()

print("Domain categories:")
print(df["Domain Applied"].unique())

print("\nApplication statuses:")
print(df["Status"].unique())

print("\nSample cleaned records:")
display(df.head(10))
```

```python
# Step 4: Clean dates and phone numbers

# Remove extra spaces and hyphens from phone numbers
df["Phone"] = (
    df["Phone"]
    .str.replace("-", "", regex=False)
    .str.replace(" ", "", regex=False)
)

# Convert dates into a consistent datetime format
dates = df["Application Date"].astype("string").str.strip()

date_format_1 = pd.to_datetime(
    dates,
    format="%Y-%m-%d",
    errors="coerce"
)

date_format_2 = pd.to_datetime(
    dates,
    format="%d/%m/%Y",
    errors="coerce"
)

df["Application Date"] = date_format_1.fillna(date_format_2)

# Display results
print("Missing dates:", df["Application Date"].isna().sum())
print("Date data type:", df["Application Date"].dtype)

print("\nSample cleaned phone numbers:")
display(df[["Phone", "Application Date"]].head(10))
```
